```python
import pandas as pd

# Load the Excel file
raw_df = pd.read_excel('/content/heart.xlsx')

# Extract the column name which contains the comma-separated header
header_line = raw_df.columns[0]
headers = header_line.split(',')

# Split each row by comma to create the clean DataFrame
# We take the first column of the raw DataFrame and split its string values
df = raw_df.iloc[:, 0].str.split(',', expand=True)
df.columns = headers

# Convert columns to appropriate numeric types where possible
df = df.apply(pd.to_numeric, errors='ignore')

display(df.head())
```
